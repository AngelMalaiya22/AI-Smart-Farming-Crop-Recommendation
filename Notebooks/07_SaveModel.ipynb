### Imports

```python
import numpy as np
import pandas as pd
import joblib
import os
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
```

### Load Dataset

```python
data1 = pd.read_csv("../DataSet/Crop_recommendation.csv")
```

```python
le = LabelEncoder()
data1['label'] = le.fit_transform(data1['label'])
```

```python
X = data1.drop('label', axis=1)
y = data1['label']

print("Training on full dataset:", X.shape)
print("Classes:", list(le.classes_))
```

### Train Final Model on Full Dataset

```python
best_params = {
    'n_estimators':      500,       
    'max_depth':         10,        
    'min_samples_leaf':  2,         
    'min_samples_split': 5,         
    'max_features':      'log2',    
    'criterion':         'log_loss',
    'random_state':      42        
}

final_model = RandomForestClassifier(**best_params)
final_model.fit(X, y)

train_acc = accuracy_score(y, final_model.predict(X))
print(f"Train Accuracy on full data: {train_acc:.4f}")
```

### Save Model and Label Encoder

```python
os.makedirs("models", exist_ok=True)

joblib.dump(final_model, "models/crop_rf_model.pkl")
joblib.dump(le,          "models/label_encoder.pkl")

print("Files saved:")
print("  models/crop_rf_model.pkl")
print("  models/label_encoder.pkl")
```

### Load and Test

```python
model_loaded = joblib.load("models/crop_rf_model.pkl")
le_loaded    = joblib.load("models/label_encoder.pkl")

# Rice growing conditions from the dataset
sample = pd.DataFrame(
    [[90, 42, 43, 20.87, 82.00, 6.50, 202.93]],
    columns=['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']
)

pred  = model_loaded.predict(sample)
label = le_loaded.inverse_transform(pred)[0]

print(f"Test Input  : N=90, P=42, K=43, Temp=20.87, Humidity=82, pH=6.5, Rainfall=202.93")
print(f"Prediction  : {label}")
print(f"Expected    : rice")
print(f"\nModel verified and ready for deployment.")
```
