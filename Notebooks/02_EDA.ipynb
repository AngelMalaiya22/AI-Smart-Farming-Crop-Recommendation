### Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

### Load the Dataset

```python
data1 = pd.read_csv("../DataSet/Crop_recommendation.csv")
```

```python
data1.head()
```

### Distribution of Nitrogen 

```python
plt.figure(figsize=(12, 6))
sb.histplot(data=data1, x='N', kde=True)  
plt.title("Distribution of Nitrogen")
plt.xlabel("Nitrogen (N)")
plt.ylabel("Count")
plt.show()
```

### Distribution of Phosphorus

```python
plt.figure(figsize=(12, 6))
sb.histplot(data=data1, x='P', kde=True)  
plt.title("Distribution of Phosphorus")
plt.xlabel("Phosphorus (P)")
plt.ylabel("Count")
plt.show()
```

### Distribution of Potassium

```python
plt.figure(figsize=(12, 6))
sb.histplot(data=data1, x='K', kde=True)  
plt.title("Distribution of Potassium")
plt.xlabel("Potassium (K)")
plt.ylabel("Count")
plt.show()
```

### Distribution of Temperature

```python
plt.figure(figsize=(12, 6))
sb.histplot(data=data1, x='temperature', kde=True)  
plt.title("Distribution of Temperature")
plt.xlabel("Temperature")
plt.ylabel("Count")
plt.show()
```

### Distribution of Humidity

```python
plt.figure(figsize=(12, 6))
sb.histplot(data=data1, x='humidity', kde=True)  
plt.title("Distribution of Humidity")
plt.xlabel("Humidity")
plt.ylabel("Count")
plt.show()
```

### Distribution of Ph

```python
plt.figure(figsize=(12, 6))
sb.histplot(data=data1, x='ph', kde=True)  
plt.title("Distribution of Ph")
plt.xlabel("Ph")
plt.ylabel("Count")
plt.show()
```

### Distribution of Rainfall 

```python
plt.figure(figsize=(12, 6))
sb.histplot(data=data1, x='rainfall', kde=True)  
plt.title("Distribution of Rainfall")
plt.xlabel("Rainfall")
plt.ylabel("Count")
plt.show()
```

### Distribution of Target- label

```python
plt.figure(figsize=(12, 6))
sb.histplot(data=data1, x='label', kde=True)  
plt.title("Distribution of label")
plt.xlabel("Target- label")
plt.ylabel("Count")
plt.show()
```

### Correlation Heatmap of Dataset

```python
plt.figure(figsize=(12,8))
sb.heatmap(
    data1.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm',
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

### Boxplots (Outlier Detection)

```python
numerical_columns = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']

for col in numerical_columns:
    plt.figure(figsize=(6,2))
    sb.boxplot(x=data1[col])
    plt.title(col)
    plt.show()
```

```python
sb.pairplot(data1[['N','P','K','temperature','humidity', 'ph', 'rainfall']])
plt.show()
```
