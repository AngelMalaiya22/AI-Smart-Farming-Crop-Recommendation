# Label Encoding
Import LabelEncoder to convert the catergorical target variable 'label' into numberical labels.

### Encoding Label
Convert each crop category into a unique numerical label for model training.

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
data1['label'] = label_encoder.fit_transform(data1['label'])
```

### Crop Label Mapping
Display the mapping between each crop name and its corresponding numerical label.

```python
print(dict(zip(label_encoder.classes_,label_encoder.transform(label_encoder.classes_))))
```
